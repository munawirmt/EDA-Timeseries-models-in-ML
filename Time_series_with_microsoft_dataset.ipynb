# **Time series operations in Microsoft dataset**

ASSIGNMENT:


*   Read microsoft dataset
*   get maximum price of the share from 2022 to 2026
*   which is the date of highest price of stock
*   which is the date of lowest price of stock





```python
import pandas_datareader as pdr
import pandas as pd
import yfinance as yf
from datetime import datetime
```

```python
df=yf.download('MSFT',start='2022-01-01',end='2026-09-19')
df
```

```python
print(df.head())
```

```python
max_price=df['High'].max()
min_price=df['Low'].min()
```

```python
max_price
```

```python
min_price
```

```python
max_date=df['High'].idxmax()
min_date=df['Low'].idxmin()
```

```python
max_date
```

```python
min_date
```

```python
print(f"Maximum high 2022-2026: ${max_price.iloc[0]:.2f} on {max_date.iloc[0].date()}")
print(f"Minimum low 2022-2026: ${min_price.iloc[0]:.2f} on {min_date.iloc[0].date()}")
```
